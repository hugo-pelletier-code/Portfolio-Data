# Analyse des statistiques de jeu des joueurs de Ligue 1

Réaliser une analyse exploratoire des données (EDA) sur le dataset des joueurs des 5 grands championnats.  
  


# À propos du dataset

Ce dataset présente les statistiques des joueurs de la saison 2024-2025 dans les cinq principaux championnats européens, issues de FBref. 

## Colonnes

### Informations de base sur le joueur
- **Player** – Nom du joueur  
- **Nation** – Nationalité  
- **Pos** – Poste (FW, MF, DF, GK)  
- **Squad** – Club  
- **Comp** – Championnat  
- **Age** – Âge du joueur  
- **Born** – Année de naissance  

### Temps de jeu et apparitions
- **MP** – Matchs joués  
- **Starts** – Titularisations  
- **Min** – Minutes jouées  
- **90s** – Nombre de matches complets (90 min)  

### Statistiques offensives
- **Gls** – Buts inscrits  
- **Ast** – Passes décisives  
- **G+A** – Buts + passes décisives  
- **xG** – Expected Goals  
- **xAG** – Expected Assists  
- **npxG** – Expected Goals sans pénalty  
- **G-PK** – Buts hors pénalty  

### Statistiques défensives
- **Tkl** – Tacles  
- **TklW** – Tacles réussis  
- **Blocks** – Dégagements (blocs)  
- **Int** – Interceptions  
- **Tkl+Int** – Tacles + interceptions  
- **Clr** – Dégagements  
- **Err** – Erreurs menant à un but  

### Statistiques de passe et créativité
- **PrgP** – Passes progressives  
- **PrgC** – Conduites de balle progressives  
- **KP** – Key Passes (passes menant à un tir)  
- **Cmp%_stats_passing** – Pourcentage de passes réussies  
- **Ast_stats_passing** – Passes décisives (autre colonne dédiée)  
- **xA** – Expected Assists  
- **PPA** – Passes dans la surface de réparation  

### Statistiques des gardiens
- **GA** – Buts encaissés  
- **Saves** – Arrêts  
- **Save%** – Pourcentage d’arrêts  
- **CS** – Clean Sheets  
- **CS%** – Pourcentage de matches sans encaisser  
- **PKA** – Pénaltys subis  
- **PKsv** – Pénaltys arrêtés  

### Possession et contrôle du ballon
- **Touches** – Touches de balle  
- **Carries** – Conduites de balle  
- **PrgR** – Courses progressives (déplacements significatifs vers l’avant)  
- **Mis** – Contrôles manqués  
- **Dis** – Pertes de balle  

### Statistiques diverses
- **CrdY** – Cartons jaunes  
- **CrdR** – Cartons rouges  
- **PKwon** – Pénaltys obtenus  
- **PKcon** – Pénaltys concédés  
- **Recov** – Ballons récupérés

# 2. Nettoyage et préparation

- Nation : ne conserver que le code pays à 3 lettres (FRA, ESP, etc.).
- Comp : supprimer le préfixe (ex. "fr Ligue 1" → "Ligue 1").
- Pos : garder le premier poste et mapper vers un nom lisible (GK → Goalkeeper, etc.).
- Filtrage : ne garder que les joueurs ayant joué au moins 17 matches (c'est à dire ayant joué au moins la moitié de la sasion) et sans valeurs manquantes pour Nation, Age ou Pos.
- Création d'une colonne "Pourcentage de pénalty arrété"

```python
football['Nation'] = football['Nation'].str.extract(r'([A-Z]{3})')

football['Comp'] = football['Comp'].str.split(' ', n=1).str[1]

football = football[football['MP'] >= 17]

football = football[football['Nation'].notna()]
football = football[football['Age'].notna()]
football = football[football['Pos'].notna()]
```

```python
def simplify_position(pos):
    first_pos = pos.split(',')[0].strip()
    if first_pos == 'GK':
        return 'Goalkeeper'
    elif first_pos == 'DF':
        return 'Defender'
    elif first_pos == 'MF':
        return 'Midfielder'
    elif first_pos == 'FW':
        return 'Forward'
    else:
        return 'Other'

football['Pos'] = football['Pos'].apply(simplify_position)
```

```python
football['PS%'] = football['PKsv'] / (football['PKA']+football['PKsv']) * 100
```

# 3. Résumé par championnat

On va comparer sur quelques valeurs clées les 5 grands championnats européens

```python
summary = (
    football.groupby('Comp')
      .agg(
         Buts                = ('Gls',       'sum'),
         Match_sans_but_concédé     = ('CS',        'sum'),
         Pénalités_concédées  = ('PKcon',     'sum'),
         Cartons_rouges       = ('CrdR',      'sum'),
         Âge_moyen            = ('Age',       'mean'),
         Tacles_interceptions = ('Tkl+Int',   'mean')
      )
      .round(2)
)

styled = (
    summary.style
           .apply(lambda s: ['color: red'   if v==s.max() else '' for v in s], axis=0)
           .apply(lambda s: ['color: green' if v==s.min() else '' for v in s], axis=0)
)

styled
```

- **Premier League**  
  - Total de buts le plus élevé (1082) et moyenne tacles + interceptions la plus forte (33,8).  
  - Cela reflète un championnat à la fois très offensif et intense au pressing, générant beaucoup d’occasions et de récupérations.

- **Serie A**  
  - Nombre de clean sheets le plus élevé (212).  
  - Confirme sa réputation de championnat tactiquement organisé et axé sur la solidité défensive.

- **La Liga**  
  - Plus grand nombre de penalties concédés (130) et de cartons rouges (78).  
  - Traduit un jeu très technique dans la surface, entraînant plus de duels perdus et d’arbitrages stricts.

- **Bundesliga**  
  - Volume de buts et de récupérations (tacles + interceptions) un peu en dessous de la PL, mais toujours élevé, illustrant le style rapide et direct allemand.

- **Ligue 1**  
  - Le plus faible total de buts (85) et le plus petit nombre de clean sheets (146).  
  - 2ᵉ plus de penalties concédés (114) et effectifs les plus jeunes (24,1 ans en moyenne).  
  - Montre un championnat où les matches sont moins prolifiques en buts, où la défense concède plus d’erreurs dans la surface, et où on mise sur des jeunes talents en plein développement.

# 5. Analyse par poste

Dans cette prtie de l'analyse on va analyser les 5 meilleurs joueurs à leur poste en ligue 1 sur des données différentes.

## Gardiens

```python
gk = ligue1[(ligue1['Pos']=='Goalkeeper') & (ligue1['MP']>=17)]

stats = ['Saves', 'Save%', 'CS', 'CS%', 'PKsv', 'PS%']
titles = ['Arrêts','% d\'arrêts','Clean sheets','% de clean sheets','Arrêts de pénalty','% d\'arrêts de pénalty']

fig = make_subplots(
    rows=3, cols=2,
    subplot_titles=titles,
    horizontal_spacing=0.25,
    vertical_spacing=0.15
)

for i, stat in enumerate(stats):
    top5 = gk.nlargest(5, stat)
    row, col = i//2 + 1, i%2 + 1
    colors = [
        'rgba(46,204,113,0.8)' if v==top5[stat].max() else 'rgba(189,195,199,0.8)'
        for v in top5[stat]
    ]
    fig.add_trace(
        go.Bar(
            y=top5['Player'],
            x=top5[stat],
            orientation='h',
            hovertext=top5['Squad'],
            marker=dict(
                color=colors,
                line=dict(color='gray', width=1.5)
            ),
            opacity=0.9
        ),
        row=row,
        col=col
    )

fig.update_layout(
    title=dict(
        text="Top 5 des gardiens – statistiques clés",
        x=0.5,
        font=dict(size=20)
    ),
    showlegend=False,
    width=1300,
    height=800,
    template='ggplot2',
    paper_bgcolor='lightgray'
)

img_bytes = fig.to_image(format='png')
display(Image(data=img_bytes))
```

```python
defense = ligue1[(ligue1['Pos']=='Defender') & (ligue1['MP']>=17)]

stats = ['Tkl', 'TklW', 'Int', 'Blocks_stats_defense', 'Tkl+Int', 'Clr']
titles = ['Tacles', 'Tacles réussis', 'Interceptions', 'Blocs', 'Tacles + Interceptions', 'Dégagements']

fig = make_subplots(rows=3, cols=2, subplot_titles=titles,
                    horizontal_spacing=0.30, vertical_spacing=0.15)

for i, stat in enumerate(stats):
    top5 = defense.nlargest(5, stat)
    r, c = divmod(i, 2)
    colors = ['rgba(46,204,113,0.8)' if v==top5[stat].max() else 'rgba(189,195,199,0.8)' 
              for v in top5[stat]]
    fig.add_trace(
        go.Bar(
            y=top5['Player'], x=top5[stat], orientation='h',
            hovertext=top5['Squad'],
            marker=dict(color=colors, line=dict(color='gray', width=1.5)),
            opacity=0.9
        ),
        row=r+1, col=c+1
    )

fig.update_layout(
    title_text="Top 5 des défenseurs – statistiques clés",
    title_x=0.5, showlegend=False,
    width=1300, height=900, template='ggplot2',
    paper_bgcolor='lightgray'
)

img_bytes = fig.to_image(format='png')
display(Image(data=img_bytes))
```

# 6. Qui mérite le Ballon d’Or : Ousmane Dembélé ou Lamine Yamal ?

```python
players = ['Lamine Yamal', 'Ousmane Dembélé']
metrics = ['Gls','Ast','G+A','Sh/90','Succ%','KP','Cmp%','Touches','Carries']

df_comp = football[football.Player.isin(players)].set_index('Player')[metrics]
df_norm = df_comp / df_comp.max()

N = len(metrics)
angles = np.linspace(0, 2*np.pi, N, endpoint=False)
angles = np.concatenate([angles, [angles[0]]])

fig, ax = plt.subplots(subplot_kw={'polar':True}, figsize=(6,6))
for player in players:
    vals = np.append(df_norm.loc[player].values, df_norm.loc[player].values[0])
    ax.plot(angles, vals, label=player)
    ax.fill(angles, vals, alpha=0.25)

ax.set_thetagrids(np.degrees(angles[:-1]), metrics)
ax.set_ylim(0,1)
ax.set_title(' Yamal vs Dembélé')
ax.legend(loc='upper left', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

En comparant les statatistiques de Yamal et Dembélé dans leur championnat respectif on se rend compte qu'ils ont deux profils très différents. Yamal est très actif au toucher de balle et en costruction de jeux, Dembélé plus précis en passe, tir et but marqué. 

Il est donc difficile de choisir le ballon d'or que sur ces données. Il fauda prendre en compte aussi celle de la ligue des champions.
